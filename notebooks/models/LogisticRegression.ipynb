# Logistic Regression
**Dataset:** UCI Optdigits

This notebook compares Logistic Regression with and without training-fold IQR outlier removal using five-fold stratified CV, then evaluates the CV-selected configuration on the untouched official test split.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

cols = [f"pixel_{i}" for i in range(64)] + ["target"]

# Load train and test separately - IMPORTANT: never fit any transformer on test data
train = pd.read_csv("../../data/processed/optdigits_train_processed.csv", header=None, names=cols)
test  = pd.read_csv("../../data/processed/optdigits_test_processed.csv", header=None, names=cols)

print("Train shape:", train.shape)   # run this code to obtain actual value
print("Test shape:", test.shape)     # run this code to obtain actual value
train.head()
```

```python
# --- Step A: total intensity per row ---
train["total_intensity"] = train.iloc[:, :64].sum(axis=1)
train["total_intensity"].describe()  # run this code to obtain actual value
```

```python
# --- Step B: IQR outlier detection (fit on TRAIN ONLY) ---
Q1 = train["total_intensity"].quantile(0.25)
Q3 = train["total_intensity"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outlier_mask = (train["total_intensity"] < lower) | (train["total_intensity"] > upper)
print("Outlier bounds:", lower, upper)          # run this code to obtain actual value
print("Number of outlier rows:", outlier_mask.sum())  # run this code to obtain actual value

train_no_outliers = train[~outlier_mask].reset_index(drop=True)
print("Rows before:", train.shape[0], "-> after:", train_no_outliers.shape[0])  # run this code to obtain actual value
```

```python
# --- EDA visualization: boxplot per digit class ---
plt.figure(figsize=(10,5))
sns.boxplot(x="target", y="total_intensity", data=train)
plt.title("Total pixel intensity per digit class (before outlier removal)")
plt.savefig("../../results/eda_visualizations/member2_outlier_boxplot.png", dpi=150, bbox_inches="tight")
plt.show()
```

## Model Evaluation

Logistic Regression is a compact multiclass baseline for the digit classes. This comparison tests whether removing train-set total-intensity IQR outliers changes performance; test rows remain untouched. MinMax scaling is fitted within each CV fold using a Pipeline, and `GridSearchCV` tunes regularization strength with 5-fold stratified CV. Accuracy and macro F1 are reported on CV and the official held-out test set; macro F1 weights every digit equally.

```python
from sklearn.base import clone
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    ConfusionMatrixDisplay,
    accuracy_score,
    classification_report,
    f1_score,
    precision_score,
    recall_score,
)
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import MinMaxScaler

feature_cols = [f"pixel_{index}" for index in range(64)]
y_test = test["target"].astype(int)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
base_splits = list(cv.split(train[feature_cols], train["target"].astype(int)))

# Estimate IQR bounds from each CV training fold and retain every validation row.
outlier_cv_splits = []
for fold_train_indices, fold_validation_indices in base_splits:
    fold_intensity = train.iloc[fold_train_indices][feature_cols].sum(axis=1)
    lower_bound, upper_bound = fold_intensity.quantile([0.25, 0.75])
    iqr = upper_bound - lower_bound
    fold_mask = fold_intensity.between(lower_bound - 1.5 * iqr, upper_bound + 1.5 * iqr).to_numpy()
    outlier_cv_splits.append((fold_train_indices[fold_mask], fold_validation_indices))

training_variants = {
    "Before IQR outlier removal": (train, base_splits),
    "After IQR outlier removal": (train, outlier_cv_splits),
}
results = []
best_models = {}
for variant_name, (cv_data, variant_cv) in training_variants.items():
    model = Pipeline([
        ("scaler", MinMaxScaler()),
        ("classifier", LogisticRegression(max_iter=2000, random_state=42)),
    ])
    is_outlier_variant = variant_name == "After IQR outlier removal"
    search = GridSearchCV(
        model,
        {"classifier__C": [0.1, 1, 10]},
        scoring={"accuracy": "accuracy", "macro_f1": "f1_macro"},
        refit=False if is_outlier_variant else "macro_f1",
        cv=variant_cv,
        n_jobs=-1,
    )
    search.fit(cv_data[feature_cols], cv_data["target"].astype(int))
    best_index = int(search.cv_results_["rank_test_macro_f1"].argmin())
    best_parameters = search.cv_results_["params"][best_index]
    if is_outlier_variant:
        final_model = clone(model).set_params(**best_parameters)
        final_model.fit(train_no_outliers[feature_cols], train_no_outliers["target"].astype(int))
    else:
        final_model = search.best_estimator_
    best_models[variant_name] = final_model
    predictions = final_model.predict(test[feature_cols])
    results.append({
        "Training data": variant_name,
        "CV accuracy": search.cv_results_["mean_test_accuracy"][best_index],
        "CV macro F1": search.cv_results_["mean_test_macro_f1"][best_index],
        "Test accuracy": accuracy_score(y_test, predictions),
        "Test precision (macro)": precision_score(y_test, predictions, average="macro"),
        "Test recall (macro)": recall_score(y_test, predictions, average="macro"),
        "Test macro F1": f1_score(y_test, predictions, average="macro"),
        "Best parameters": best_parameters,
    })

outlier_model_comparison = pd.DataFrame(results).sort_values(
    "CV macro F1", ascending=False
).reset_index(drop=True)
display(outlier_model_comparison)

best_variant = outlier_model_comparison.loc[0, "Training data"]
best_predictions = best_models[best_variant].predict(test[feature_cols])
print(f"CV-selected model: {best_variant}")
print("Classification report on the held-out official test split:")
print(classification_report(y_test, best_predictions, digits=4))
ConfusionMatrixDisplay.from_predictions(y_test, best_predictions, cmap="Blues", colorbar=False)
plt.title(f"Held-out test confusion matrix: {best_variant}")
plt.tight_layout()
plt.show()
```

### Results and conclusion

The train-only IQR rule removed 3 of 3,823 training examples. After-removal CV macro F1 was 97.18%, slightly above 97.12% before removal, but test macro F1 was 95.22%, slightly below 95.28% before removal. The effect is small and mixed, so this experiment does not show that outlier removal improves unseen-data performance. The official test rows were retained unchanged, and scaling was fitted within CV folds. Limitations include using total intensity as a one-dimensional outlier rule and evaluating on one official test split; future work could inspect the removed examples and compare robust models or alternative outlier definitions.
